## 1. Scatter Plot

### Axes-Level Function

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='time',
    size='size'
)
```

### Figure-Level Function

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    kind='scatter',
    hue='sex'
)
```

## Important Parameters

- **`data`** → The DataFrame has to be passed.
  - Example: `data=tips`
  - The DataFrame itself should **not** be inside quotes.

- **`x` and `y`** → Specify the columns to plot on the X-axis and Y-axis.

- **`hue`** → Separates the data using different **colors** according to a column and automatically creates a legend.

- **`size`** → Changes the **size of the points** according to the values/categories of a column.

- **`style`** → Changes the **marker style** of the points according to the categories of a column.

- Other parameters such as **`color`**, **`marker`**, and **`linestyle`** can also be passed.

## `scatterplot()` vs `relplot()`

`sns.scatterplot()` is an **axes-level function** specifically used for scatter plots.

`sns.relplot()` is a general **figure-level relational plotting function**, so we specify:

```python
kind='scatter'
```

to tell Seaborn that we want a scatter plot.

## Axes-Level vs Figure-Level

**Axes-level → `sns.scatterplot()`**

Works on a single Matplotlib `Axes` object and controls only the space allocated to that axes.

**Figure-level → `sns.relplot()`**

Creates and manages the **entire figure** using Seaborn's `FacetGrid`.

Because it controls the entire figure, it can also easily create multiple plots using parameters such as **`row`** and **`col`**.

> **Legend:** Figure-level functions often place the legend **outside the plotting axes** because they can use the entire graph, while axes-level functions usually place the legend **inside the axes** by default because they control the space allocated to them. The exact position can vary.

# 3) Facetplot

Facetplot only works with **figure-level functions** because it creates multiple graphs depending upon the categories, so it needs access to the entire figure.\
\
Faceting means splitting one plot into multiple smaller plots based on categories in your data.
### Using `row` and `col`
```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='smoker',
    row='sex',
    kind='line'
)
```

* `col='smoker'` -> creates columns based on smoker categories
* `row='sex'` -> creates rows based on sex categories
* Here it creates **4 plots**:
  * Male + Smoker
  * Male + Non-Smoker
  * Female + Smoker
  * Female + Non-Smoker

### Using only `col`
```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='day',
    kind='line'
)
```

* `col='day'` -> creates one plot for each unique day
* This gives **4 plots side-by-side**
* If there are many categories, the plots can become cluttered

### Using `col_wrap`
```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='day',
    col_wrap=2,
    kind='line'
)
```

* `col_wrap=<int>` -> controls how many plots are displayed in one row
* `row` -> creates facets vertically
* `col` -> creates facets horizontally
* `col_wrap` -> wraps `col` facets into multiple rows
* `row` and `col_wrap` **cannot be used together**

# sns.displot()

Figure-level function for visualizing the **distribution of data**.

### Basic Syntax
```python
sns.displot(
    data=None,
    x=None,
    y=None,
    hue=None,
    row=None,
    col=None,
    weights=None,
    kind='hist',
    rug=False,
    rug_kws=None,
    log_scale=None,
    legend=True,
    palette=None,
    hue_order=None,
    hue_norm=None,
    color=None,
    col_wrap=None,
    row_order=None,
    col_order=None,
    height=5,
    aspect=1,
    facet_kws=None,
    **kwargs
)
```

## Parameters

* `data` -> DataFrame containing the data
* `x` -> variable plotted on the x-axis
* `y` -> variable plotted on the y-axis; mainly used for bivariate distributions
* `hue` -> divides data into groups using different colors
* `row` -> creates separate subplot rows based on a categorical variable
* `col` -> creates separate subplot columns based on a categorical variable
* `weights` -> gives different observations different importance/weight
* `kind` -> selects the type of distribution plot:
  * `kind='hist'` -> Histogram
  * `kind='kde'` -> KDE (Kernel Density Estimate)
  * `kind='ecdf'` -> ECDF (Empirical Cumulative Distribution Function)
* `rug=True` -> adds small marks representing individual observations
* `rug_kws` -> dictionary containing customization options for the rug plot
* `log_scale` -> uses logarithmic scaling on the axis
* `legend` -> controls whether the legend is displayed
* `palette` -> controls colors when `hue` is present
* `hue_order` -> controls the order of categories used by `hue`
* `hue_norm` -> controls color normalization when `hue` is numerical
* `color` -> sets one color for the plot
* `col_wrap` -> maximum number of subplot columns before wrapping onto another row
* `row_order` -> controls which row categories appear and their order
* `col_order` -> controls which column categories appear and their order
* `height` -> height of **each subplot/facet** in inches
* `aspect` -> width-to-height ratio of each subplot

```text
width = height × aspect
```

* `facet_kws` -> dictionary containing additional settings for the underlying `FacetGrid`
* `**kwargs` -> additional parameters passed to the underlying plotting function\
\
\
\
---
# Histogram Parameters

When:

```python
kind='hist'
```

`displot()` uses `histplot()` underneath.

Therefore, `histplot()` parameters can also be passed.

* `bins` -> number of bins or exact bin boundaries
* `binwidth` -> width of each bin
* `binrange` -> minimum and maximum values covered by the bins
* `discrete` -> treats values as discrete rather than continuous
* `stat` -> determines what the height of the histogram represents:
  * `stat='count'` -> number of observations
  * `stat='frequency'` -> observations divided by bin width
  * `stat='probability'` -> proportion of observations
  * `stat='percent'` -> percentage of observations
  * `stat='density'` -> probability density
* `cumulative=True` -> values accumulate from left to right
* `common_bins` -> whether `hue` groups use the same bins
* `common_norm` -> whether normalization is shared between `hue` groups
* `multiple` -> controls how distributions created using `hue` are displayed:
  * `multiple='layer'` -> histograms overlap
  * `multiple='dodge'` -> histograms appear side-by-side
  * `multiple='stack'` -> histograms are stacked
  * `multiple='fill'` -> histograms are stacked and normalized to proportions
* `element` -> controls how the histogram is drawn:
  * `element='bars'` -> normal rectangular bars (default)
  * `element='step'` -> step/outline shape
  * `element='poly'` -> connects bin heights into a polygon-like shape
* `fill` -> controls whether the histogram element is filled
* `shrink` -> shrinks the width of histogram bars
* `thresh` -> minimum value required before a bin is displayed
* `pthresh` -> threshold expressed as a proportion
* `pmax` -> controls saturation of the color mapping for bivariate histograms
* `cbar=True` -> adds a colorbar for a bivariate histogram
* `cbar_ax` -> specifies the Axes where the colorbar should be drawn
* `cbar_kws` -> additional settings for the colorbar

# rugplot()

Shows **every individual observation** in the dataset as a small tick/line along an axis.

### Basic Syntax
```python
sns.rugplot(
    data=tips,
    x='total_bill'
)
```

### Example Data
```python
x = [10, 12, 12, 15, 20]
```

A rug plot would look roughly like:

```text
10    12       15              20
|     ||       |               |
─────────────────────────────────
```

* Each `|` represents **one actual observation**
* Observations are **not grouped or aggregated**
* If multiple observations are close together, the ticks become dense

```text
| |||||| |||      |   |
```

* Dense ticks -> many observations exist in that region
* Sparse ticks -> fewer observations exist in that region

---

# Combining Rugplot with Other Plots

# KDE Parameters

When:

```python

```

`displot()` uses `kdeplot()` underneath.

* `bw_method` -> controls how the KDE bandwidth is calculated
* `bw_adjust` -> adjusts the bandwidth:
  * smaller value -> more detailed/wavy curve

* `fill=True` -> fills the area underneath the KDE curve
* `multiple` -> controls how multiple KDE distributions interact
* `common_norm` -> controls whether normalization is shared between `hue` groups

* `clip` -> limits the range over which the KDE is calculated
* `gridsize` -> number of points used to calculate the KDE curve
* `cumulative=True` -> produces a cumulative KDE

# 1) heatmap()

A heatmap represents values in a **2D table/matrix using colors**.

Each cell represents one value, and the color of the cell tells us how large or small that value is.

### Basic Syntax

```python
sns.heatmap(
    data,
    annot=False,
    fmt='.2g',
    cmap=None,
    vmin=None,
    vmax=None,
    center=None,
    robust=False,
    linewidths=0,
    linecolor='white',
    square=False,
    xticklabels='auto',
    yticklabels='auto',
    cbar=True,
    cbar_kws=None,
    mask=None,
    ax=None
)
```

### Simple Example

```python
data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data)
```

Conceptually:

```text
Smaller value
     ↓
one end of color scale

Larger value
     ↓
other end of color scale
```

# 2) clustermap()

A clustermap is essentially:

```text
Heatmap + Hierarchical Clustering
```

It displays a heatmap while also **reordering similar rows and/or columns so that similar observations appear near each other**.

```python
sns.clustermap(data)
```

Unlike a normal heatmap:

```text
heatmap()
→ Displays the matrix in its existing order

clustermap()
→ Finds similarity
→ Groups similar rows/columns
→ Reorders them
→ Displays the reordered heatmap
```

---

# Dendrogram

A clustermap adds a **dendrogram**.

A dendrogram is the tree-like structure displayed beside the heatmap.

```text
        ┌── A
    ┌───┤
    │   └── B
────┤
    │   ┌── C
    └───┤
        └── D
```

If `A` and `B` join very early:

```text
A ─┐
   ├─
B ─┘
```

they are relatively similar according to the chosen distance/linkage method.

So:

```text
Dendrogram
     ↓
Shows hierarchical similarity/grouping

Heatmap
     ↓
Shows the actual values
```

---

# Basic Example

```python
sns.clustermap(
    data
)
```

The data should generally contain **numerical values** for the matrix being clustered.

For a DataFrame:

```python
numeric_df = tips.select_dtypes(include='number')

sns.clustermap(
    numeric_df

```

# Common Parameters

```python
sns.clustermap(

    metric='euclidean',
    cmap='viridis',
    standard_scale=None,
    z_score=None,
    row_cluster=True,
    col_cluster=True,
    figsize=(10, 8)
)
```

* `data` -> 2D numerical matrix

* `cmap` -> controls heatmap colors

* `method` -> determines how clusters are combined during hierarchical clustering

Common examples:

```python
method='single'
method='complete'
method='average'
method='ward'
```

* `metric` -> determines how distance/similarity between observations is calculated

Common example:

```python
metric='euclidean'
```

* `row_cluster=True` -> clusters and reorders rows

* `col_cluster=True` -> clusters and reorders columns

To disable one:

```python
sns.clustermap(
    data,
    row_cluster=True,
    col_cluster=False
)
```

# Standardizing Data

Clustering is based on distances, so variables with very different scales can strongly affect the result.

For example:

```text
Age       → 18 - 70
Income    → 20,000 - 2,000,000
```

Income may dominate the distance calculation.

`clustermap()` provides options for scaling.

### standard_scale

```python
sns.clustermap(
    data,
    standard_scale=1
)
```

* rescales values relative to the selected axis

### z_score

```python
sns.clustermap(
    data,
    z_score=1
)
```

* standardizes values using z-scores

Conceptually:

```text
Different scales
      ↓
Standardization
      ↓
Comparable scales
      ↓
More meaningful distance calculation
```

## pivot() vs pivot_table()

### pivot()

`pivot()` is used to **reshape data**.

```python
df.pivot(
    index='country',
    columns='year',
    values='lifeExp'
)
```

* Does **not** perform aggregation
* Each `(index, column)` combination must be **unique**
* If duplicate combinations exist -> error

```text
country       year      lifeExp
India         2000        62
India         2000        65
```

Here:

```python
df.pivot(
    index='country',
    columns='year',
    values='lifeExp'
)
```

will give:

```text
ValueError: Index contains duplicate entries, cannot reshape
```

because Pandas doesn't know which value should go here:

```text
             2000
India         ???
```

---

### pivot_table()

`pivot_table()` can handle duplicate combinations because it **aggregates them**.

```python
df.pivot_table(
    index='country',
    columns='year',
    values='lifeExp',
    aggfunc='mean'
)
```

For:

```text
India    2000    62
India    2000    65
```

it calculates:

```text
(62 + 65) / 2 = 63.5
```

and produces:

```text
             2000
India        63.5
```

* Default `aggfunc='mean'`
* Can be changed:

```python
aggfunc='sum'
aggfunc='median'
aggfunc='min'
aggfunc='max'
aggfunc='count'
```

### Comparison

```text
                    pivot()          pivot_table()

Duplicates             ✕                   ✓
Aggregation            ✕                   ✓
Default aggregation    —                  mean
Reshaping              ✓                   ✓
Flexibility           less                more
```

# Aggregation in lineplot() / relplot()

When multiple `y` observations exist for the same `x`, Seaborn's **line plot** estimates a representative value for that `x`.

Example:

```text
BP      C
5       1
5       2
5       3
```

Using:

```python
sns.lineplot(
    data=df,
    x='BP',
    y='C'
)
```

The default estimator is the **mean**.

```text
C mean = (1 + 2 + 3) / 3
       = 2
```

So the estimated point is:

```text
BP = 5
C  = 2
```

Conceptually:

```text
(5,1) ─┐
(5,2) ─┼──→ mean ──→ (5,2)
(5,3) ─┘
```

---

# Changing the Estimator

The default is:

```python
estimator='mean'
```

It can be changed.

```python
sns.lineplot(
    data=df,
    x='BP',
    y='C',
    estimator='median'
)
```

You can also pass functions:

```python
sns.lineplot(
    data=df,
    x='BP',
    y='C',
    estimator=np.sum
)
```

```python
sns.lineplot(

    x='BP',
    y='C',

    errorbar=None
)
```

* `estimator=None` -> don't aggregate repeated `x` values
* `errorbar=None` -> don't display the uncertainty/error interval

# Error / Confidence Region

By default, `lineplot()` can show an error band around the estimated line.

```text
      ╱ shaded region ╲
─────╱──── line ───────╲─────
```

The line represents the **estimated central value**.

The shaded region represents **uncertainty around that estimate**.

It is not simply saying that the observations contain "errors".

To remove it:

```python
sns.lineplot(
    data=df,
    x='BP',
    y='C',
    errorbar=None
)
```

# 1) boxplot()

A boxplot summarizes the distribution of numerical data using **quartiles, median, whiskers and outliers**.

# Structure of a Boxplot

```text
                  outlier
                     •

       whisker
          |
          |
      ┌─────────┐  ← Q3 (75th percentile)
      │         │
      │─────────│  ← Median (50th percentile)
      │         │
      └─────────┘  ← Q1 (25th percentile)
          |
          |
       whisker
```

Important values:

```text
Q1 = 25th percentile
Q2 = 50th percentile = Median
Q3 = 75th percentile
```

The box contains the **middle 50% of the data**.

```text
Q1 ├───────────────┤ Q3
        50%
```

---

# IQR

The size of the box is called the **Interquartile Range (IQR)**.

```text
IQR = Q3 - Q1
```

It represents the spread of the **middle 50% of observations**.

---

# Whiskers

The whiskers do NOT necessarily represent the actual minimum and maximum.

By default, they usually extend to the most extreme observations within:

```text
Lower limit = Q1 - 1.5 × IQR

Upper limit = Q3 + 1.5 × IQR
```

Values beyond these limits are normally displayed individually as **outliers**.

```text
•       |────[=====|=====]────|        •
↑                                      ↑
outlier                              outlier
```

So a boxplot gives information about:

```text
→ Median
→ Q1
→ Q3
→ IQR
→ Spread
→ Whiskers
→ Potential outliers
```

---

```python
    x='day',
    y='total_bill',
    hue='sex'
)
```

* `hue` -> divides each category into additional groups

# Error Bars and Bootstrapping

Like `barplot()`, `pointplot()` can display uncertainty around the estimate.

```text
        │
        ●
        │
```

With the default confidence-interval error bars, Seaborn uses **bootstrapping to estimate the confidence interval**, not to calculate the mean itself.

```text
Original observations
        ↓
Calculate mean
        ↓
Bootstrap resampling
        ↓
Estimate uncertainty of mean
        ↓
Confidence interval
```

So this statement:

```text
"Pointplot performs bootstrapping to calculate the mean"
```

is not quite correct.

More accurately:

```text
Pointplot calculates the mean directly
        +
Bootstrapping can be used to estimate its confidence interval
```

To remove error bars:

```python
sns.pointplot(
    data=tips,
    x='day',
    y='total_bill',
    errorbar=None
```

# 3) residplot()

`residplot()` plots the **residuals/errors of a regression model**.

```python

    data=tips,
    x='total_bill',
    y='tip'
)
```

---

# What is a Residual?

For every observation:

```text
Residual = Actual value - Predicted value
```

Example:

```text
Actual tip    = 5
Predicted tip = 4

Residual = 5 - 4
         = 1
```

Another:

```text
Actual tip    = 3
Predicted tip = 4

Residual = 3 - 4
         = -1
```

Conceptually:

```text
Regression:

Actual point      •
                  │ ← residual
                  │
Regression line ──●────────
               predicted
```

---

# Residual Plot

Instead of plotting the original `y` values, `residplot()` plots:

```text
x vs residual
```

Conceptually:

```text
Residual

 +2 │      •
 +1 │ •          •
  0 ├────────────────────
 -1 │    •           •
 -2 │         •
    └──────────────────── x
```

The horizontal line at:

```text
residual = 0
```

means:

```text
Actual = Predicted
```

Points above zero:

```text
Residual > 0
→ Actual > Predicted
→ Model underpredicted
```

Points below zero:

```text
Residual < 0
→ Actual < Predicted
→ Model overpredicted
```

---

# Why Use residplot()?

Residual plots help us check whether a regression model is behaving appropriately.

Ideally:

```text
•   •      •
   •   •
────────────── 0
 •      •   •
    •
```

Residuals should generally look **randomly scattered around 0**.

If there is a clear pattern:

```text
•                 •
  •             •
    •         •
      •     •
────────•──────────
```

it may suggest that the fitted relationship is missing some structure in the data.

---

# Mental Model

```text
regplot()
→ What relationship did the regression model fit?


residplot()
→ Where and by how much is that fitted model wrong?
```

It is a lower-level Seaborn interface that gives more control over **faceting**.

# map() vs map_dataframe()

It is a quick, high-level way of exploring relationships between multiple variables.

# Bivariate KDE

Instead of displaying individual observations as dots, KDE estimates **where observations are concentrated**.

```text
More observations in an area
            ↓
      Higher density

Fewer observations in an area
            ↓
       Lower density
```

The contour lines represent regions of similar estimated density.

Conceptually:

```text
        __________
      /            \
     /    ______    \
    /    /      \    \
   |    |  ____  |    |
   |    | |    | |    |
    \    \______/    /
     \              /
      \____________/
```

Moving toward the dense inner regions generally means **higher estimated density**.

Because:

the distributions are calculated/displayed separately for the different species.
